```python
import requests
import pandas as pd
import time
# =========================
# OpenAlex API
# =========================
base_url = "https://api.openalex.org/works"

# =========================
# 查询参数
# =========================
params = {

    # 检索标题和摘要中的 ADHD
    "search.title_and_abstract": "ADHD",

    # 文献过滤条件
    "filter": (
        "type:article,"
        "publication_year:2023-2025,"
        "primary_topic.field.id:27"
    ),

    # 每页最大200
    "per-page": 200,

    # 游标翻页
    "cursor": "*"
}

# =========================
# 保存数据
# =========================
all_papers = []

print("开始下载 ADHD 文献数据...")

# =========================
# 开始循环下载
# =========================
while True:

    try:

        response = requests.get(base_url, params=params)

        # 请求失败
        if response.status_code != 200:

            print("请求失败：", response.status_code)
            break

        result = response.json()

        page_items = result.get("results", [])

        # 没有数据则结束
        if not page_items:
            break

        # =========================
        # 遍历每篇论文
        # =========================
        for item in page_items:

            # ---------------------------------
            # 摘要恢复
            # ---------------------------------
            abstract = ""

            inverted_index = item.get(
                "abstract_inverted_index"
            )

            if inverted_index:

                word_list = []

                for word, positions in inverted_index.items():

                    for pos in positions:

                        word_list.append((pos, word))

                # 按位置排序
                word_list.sort()

                # 拼接摘要
                abstract = " ".join(
                    [word for pos, word in word_list]
                )

            # ---------------------------------
            # 作者、机构、国家
            # ---------------------------------
            authors = []
            institutions = []
            countries = []

            authorships = item.get("authorships")

            if authorships:

                for auth in authorships:

                    # 作者
                    author_info = auth.get("author")

                    if (
                        author_info
                        and author_info.get("display_name")
                    ):

                        authors.append(
                            str(
                                author_info.get(
                                    "display_name"
                                )
                            )
                        )

                    # 机构
                    insts = auth.get("institutions")

                    if insts:

                        for inst in insts:

                            # 机构名称
                            inst_name = inst.get(
                                "display_name"
                            )

                            if inst_name:

                                institutions.append(
                                    str(inst_name)
                                )

                            # 国家
                            country = inst.get(
                                "country_code"
                            )

                            if country:

                                countries.append(
                                    str(country)
                                )

            # 去重
            authors = list(set(authors))
            institutions = list(set(institutions))
            countries = list(set(countries))

            # ---------------------------------
            # Concepts
            # ---------------------------------
            concepts = []

            concept_data = item.get("concepts")

            if concept_data:

                concepts = [

                    str(c.get("display_name"))

                    for c in concept_data

                    if c.get("display_name")
                ]

            # ---------------------------------
            # Topics
            # ---------------------------------
            topics = []

            topic_data = item.get("topics")

            if topic_data:

                topics = [

                    str(t.get("display_name"))

                    for t in topic_data

                    if t.get("display_name")
                ]

            # ---------------------------------
            # 期刊
            # ---------------------------------
            journal = ""

            primary_location = item.get(
                "primary_location"
            )

            if (
                primary_location
                and primary_location.get("source")
            ):

                source = primary_location.get("source")

                if source:

                    journal_name = source.get(
                        "display_name"
                    )

                    if journal_name:

                        journal = str(journal_name)

            # ---------------------------------
            # 开放获取
            # ---------------------------------
            is_oa = False

            open_access = item.get("open_access")

            if open_access:

                is_oa = open_access.get(
                    "is_oa",
                    False
                )

            # ---------------------------------
            # 保存结构化数据
            # ---------------------------------
            all_papers.append({

                # 基础信息
                "paper_id": item.get("id"),

                "title": item.get("title"),

                "abstract": abstract,

                "publication_year": item.get(
                    "publication_year"
                ),

                "cited_by_count": item.get(
                    "cited_by_count"
                ),

                "doi": item.get("doi"),

                # 期刊
                "journal": journal,

                # OA
                "is_oa": is_oa,

                # 作者
                "authors": "; ".join(authors),

                "authors_count": len(authors),

                # 机构
                "institutions": "; ".join(
                    institutions
                ),

                "institutions_count": len(
                    institutions
                ),

                # 国家
                "countries": "; ".join(countries),

                "countries_count": len(countries),

                # 概念
                "concepts": "; ".join(concepts),

                # 主题
                "topics": "; ".join(topics)
            })

        # 输出下载进度
        print(f"已下载：{len(all_papers)} 条")

        # 下一页游标
        next_cursor = result.get(
            "meta",
            {}
        ).get("next_cursor")

        # 没有下一页
        if not next_cursor:
            break

        # 更新游标
        params["cursor"] = next_cursor

        # 防止请求过快
        time.sleep(1)

    except Exception as e:

        print("发生错误：", e)
        break

# =========================
# 转为DataFrame
# =========================
df = pd.DataFrame(all_papers)

# =========================
# 删除重复 DOI
# =========================
df = df.drop_duplicates(
    subset="doi"
)

# =========================
# 删除标题为空
# =========================
df = df.dropna(
    subset=["title"]
)

# =========================
# 保存 CSV
# =========================
df.to_csv(
    "ADHD_OpenAlex数据集.csv",
    index=False,
    encoding="utf-8-sig"
)

# =========================
# 输出结果
# =========================
print("\n数据下载完成！")

print(f"最终文献数：{len(df)}")

print(
    "文件已保存：ADHD_OpenAlex数据集.csv"
)
```
